```python
import pandas as pd
import yfinance as yf

spyticker = yf.Ticker("SPY")
df_ticker = spyticker.history(period="max", interval="1d", start="2000-12-01", end="2026-01-01" , auto_adjust=False, rounding=True)
df_ticker.head(5)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()

ax.plot(df_ticker.index, df_ticker["Adj Close"])

ax.set(xlabel='time', ylabel='price',
       title='SPY')

ax.grid()

plt.show()
```

```python
daily_return = df_ticker["Close"].pct_change(1)
daily_return_clean = daily_return.dropna()
daily_return_clean = daily_return_clean * 100 

daily_return_clean
```
